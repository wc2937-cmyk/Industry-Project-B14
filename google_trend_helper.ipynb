```python
# Install in a notebook / terminal, not inside .py file:
# !pip install pytrends
# %pip install --upgrade pip

from pytrends.request import TrendReq
import pandas as pd

class GoogleTrendHandler:
    def __init__(self, keywords, start_date="2020-01-01", end_date="2025-01-01", geo="US"):
        self.keywords = keywords
        self.start_date = start_date
        self.end_date = end_date
        self.geo = geo
        self.trends = None

    def load_trends(self, save_path="googletrend_data.xlsx"):
        pytrends = TrendReq()
        # timeframe controls the period
        pytrends.build_payload(
            self.keywords,
            timeframe=f"{self.start_date} {self.end_date}",
            geo=self.geo
        )
        self.trends = pytrends.interest_over_time()

        if "isPartial" in self.trends.columns:
            self.trends = self.trends.drop(columns=["isPartial"])

        # forward-fill any gaps
        self.trends = self.trends.fillna(method="ffill")

        # save to Excel
        self.trends.to_excel(save_path)
        print(f"📈 Google Trends data saved to {save_path}")
        return self.trends


if __name__ == "__main__":
    # choose your keywords and time period here
    keywords = ["Apple stock"]
    start_date = "2020-01-01"
    end_date = "2025-01-01"

    trends_handler = GoogleTrendHandler(
        keywords=keywords,
        start_date=start_date,
        end_date=end_date,
        geo="US"
    )

    trends_df = trends_handler.load_trends("googletrend_data.xlsx")
```
